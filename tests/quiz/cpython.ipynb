## Введение в Python и CPython

**1. Что такое CPython и чем он отличается от Python?**

Python — язык программирования с определёнными синтаксисом и семантикой.
CPython — его основная, наиболее распространённая реализация: программа,
которая выполняет код на Python. Ядро CPython написано на C.

**2. Ответьте на вопрос с изображения в задании.**

На изображении сравниваются язык Python и его реализация CPython:

| Аспект | Python | CPython |
| --- | --- | --- |
| Определение | Язык высокого уровня | Основная реализация Python |
| Исполнение | Правила языка | Компиляция в байт-код и его выполнение |
| Реализация | Не фиксирует язык реализации | Ядро написано на C |
| Примеры | Задаёт `for`, `if`, функции | Выполняет такой код |

CPython компилирует исходный Python-код в байт-код, который выполняет
его виртуальная машина. Поэтому `for` и `if` — конструкции языка Python,
а CPython обеспечивает их выполнение. При этом не весь проект написан
на C: значительная часть стандартной библиотеки написана на Python.

**3. Сколько существует реализаций Python и какая самая популярная?**

Фиксированного числа нет: реализации появляются и развиваются.
Примеры: CPython, PyPy, Jython, IronPython, MicroPython и GraalPy.
Это не полный список. Самая распространённая реализация — CPython.

**4. На каком языке написан CPython?**

Ядро интерпретатора написано на C. Значительная часть стандартной
библиотеки написана на самом Python.

## Поиск и установка CPython

**5. (Опционально) Кто создал CPython?**

Гвидо ван Россум (Guido van Rossum).

**6. Почему Python считается быстрым, несмотря на то, что это
интерпретируемый язык?**

Нужно различать скорость разработки и скорость выполнения программы.
Python удобен для быстрой разработки, но обычные циклы в CPython часто
медленнее аналогичного кода на C. Встроенные операции и библиотеки могут
выполнять основную работу в оптимизированном нативном коде. Поэтому
скорость зависит от задачи и используемых библиотек.

## Структура CPython

**8. Что содержится в папке include в CPython?**

В дереве исходников каталог называется `Include`. Он содержит
заголовочные файлы C (`.h`): публичный C API, детали API CPython
и внутренние заголовки. Это не то же самое, что исходники `.c`.
В установленном Python путь к заголовкам зависит от сборки и окружения;
его можно получить через `sysconfig.get_path("include")`.

**9. Где найти исходный код CPython?**

https://github.com/python/cpython

**10. (Опционально) Как работает CPython при выполнении кода?**

CPython разбирает исходный код и компилирует его в промежуточный
байт-код. Затем байт-код выполняет виртуальная машина Python (PVM),
то есть интерпретатор байт-кода внутри CPython. Это не машинные
инструкции, которые процессор исполняет напрямую.

## Запуск файла с помощью CPython

**11. Какая команда используется для запуска файла через CPython?**

`python example.py` или `python3 example.py`, если выбранная команда
указывает на CPython. Команда без имени файла открывает интерактивный
интерпретатор. Проверка реализации:

```bash
python -c "import platform; print(platform.python_implementation())"
```

**12. Можно ли запускать текстовые файлы через интерпретатор Python?**

Да, например `python example.txt`, если файл содержит корректный код
Python. При явном запуске через интерпретатор расширение файла
не определяет язык; содержимое должно быть кодом Python.

**13. Как указать путь к интерпретатору и файлу для выполнения кода?**

Сначала указывают исполняемый файл интерпретатора, затем файл программы.
Пути с пробелами заключают в кавычки. Пример для Linux:

```bash
"/path/to/python" "/path/to/example.py"
```

## Введение в PyPy

**14. Чем PyPy отличается от CPython?**

PyPy — другая реализация Python; её интерпретатор написан на RPython,
ограниченном подмножестве Python. В обычной сборке PyPy есть
JIT-компилятор (Just-In-Time): во время работы он переводит часто
выполняемые участки кода в машинные инструкции.
CPython 3.11 выполняет байт-код специализирующим интерпретатором без JIT.
Утверждение «в CPython никогда нет JIT» неверно: в новых версиях
доступны сборки с экспериментальным JIT.

**15. Почему PyPy подходит не для всех проектов на Python?**

Нужно учитывать поддерживаемую версию Python и совместимость зависимостей.
Расширения, использующие C API CPython, могут требовать отдельной сборки
для PyPy и работать с накладными расходами. Для коротких программ
прогрев JIT может не окупиться, а код внутри нативных библиотек JIT
PyPy обычно не ускоряет. Совместимость и скорость проверяют на проекте.

**16. Где можно скачать PyPy?**

https://www.pypy.org/download.html

## Установка и запуск PyPy

**17. Как установить PyPy после скачивания?**

Скачать сборку под свою ОС и архитектуру, распаковать архив целиком
и запустить интерпретатор из него. В Linux исполняемый файл находится
в каталоге `bin`; его точное имя зависит от сборки. Можно добавить этот
каталог в `PATH` или создать символическую ссылку. Переносить только
исполняемый файл нельзя: ему нужны библиотеки из распакованного архива.

**18. Как запустить файл с помощью PyPy?**

`pypy3 example_pypy.py`, если команда `pypy3` доступна в `PATH`.
Иначе нужно указать фактический путь к исполняемому файлу PyPy.
Имя `pypy3.12` подходит только для сборки с таким именем файла.

**19. Почему PyPy может выполнять код быстрее CPython?**

JIT компилирует и оптимизирует часто выполняемые участки кода, сокращая
затраты на интерпретацию. Это может ускорять вычислительные циклы
на чистом Python. Ускорение не гарантировано для любой программы:
оно зависит от задачи, прогрева JIT, версии и используемых библиотек.

### Задание 2. Исследование структуры CPython

Исходный ответ содержал результат «163 файла на C». Однако команда
с маской `*.h` считала заголовки, а не исходники `.c`. Кроме того,
`sys.base_prefix + '/include'` может содержать заголовки других библиотек
и не совпадать с каталогом заголовков текущего Python.
Поэтому число 163 нельзя считать подтверждённым числом файлов CPython.

Для проверки именно заголовков текущего интерпретатора:

```python
import sysconfig
from pathlib import Path

include_dir = Path(sysconfig.get_path("include"))
print("Include:", include_dir)
print("Headers (*.h):", sum(p.is_file() for p in include_dir.glob("*.h")))
```

Команда считает файлы `.h` только в выбранном каталоге, без подкаталогов.
Для рекурсивного подсчёта нужна `rglob("*.h")`; это уже другой подсчёт.
Конкретное число зависит от версии и сборки Python.
Контрольная проверка 28.10.2026 в отдельной среде Linux с CPython 3.11.16
дала 73 файла `.h` на верхнем уровне и 189 с подкаталогами; файлов `.c`
на верхнем уровне нет. Это результат проверочной среды, не компьютера
автора и не окружения Conda `ds-course`.
Описание проекта находится в `README.rst` репозитория:
https://github.com/python/cpython

### Задание 5. Сравнение производительности CPython и PyPy

Код `performance_test.py` из задания сохранён с исходным таймером,
чтобы записанные ниже результаты относились к тому же примеру:

```python
import time

start_time = time.time()
total = 0
for i in range(1, 10_000_000):
    total += i
end_time = time.time()

print("Result:", total)
print("Execution time:", end_time - start_time, "seconds")
```

Команды запуска:

```bash
python performance_test.py
pypy3 performance_test.py
```

Исходные результаты из PR (по одному замеру для каждого интерпретатора):

| Интерпретатор | Result | Время, с |
| --- | --- | --- |
| CPython (`python3`) | 49999995000000 | 0.7251119613647461 |
| PyPy (`pypy3`) | 49999995000000 | 0.10695767402648926 |

`range(1, 10_000_000)` не включает 10 000 000. Проверка суммы:
`9_999_999 * 10_000_000 // 2 = 49_999_995_000_000`.
Оба записанных результата верны.

Отношение времён: `0.7251119613647461 / 0.10695767402648926`,
то есть примерно **6,78 раза** (округлённо — 7 раз) в этом запуске.
Такой результат согласуется с оптимизацией цикла JIT-компилятором PyPy,
но один замер не доказывает, что PyPy всегда быстрее в семь раз.

Для исходных замеров не сохранены полные версии интерпретаторов,
характеристики машины, нагрузка и повторные запуски. Их точное время
нельзя воспроизвести только по этим двум числам.
Таймер охватывает цикл, но не запуск процесса. Время прогрева JIT
внутри цикла включено; отдельного прогрева перед замером нет.

Для повторной проверки нужно сохранить вывод `python --version`
и `pypy3 --version`, модель CPU и ОС, выполнить один и тот же файл
минимум три раза каждым интерпретатором на одной машине и сравнить
медианы. Запуски из новых процессов включают прогрев JIT каждый раз.
Измерение уже прогретого кода внутри одного процесса — отдельный опыт.
Для новых измерений длительности предпочтителен `time.perf_counter()`;
после смены таймера времена нужно измерить заново для обоих вариантов.
